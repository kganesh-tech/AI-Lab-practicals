```python
room=["D","C","D","D","C"]
def show_room(room):
    picture=""
    for spot in room:
        if spot=="D":
            picture+="💩"
        else:
            picture+="✨"
    print(picture)
print("our room is:")
show_room(room)
```

```python
def clean_spot(spot):
    if spot=="D":
        return "C"
    else:
        return "C"
result=clean_spot("D")
print("the robot looked at the dirty part and made it:",result,"(C means clean✨)")
```

```python
print("BEFORE-the dirty room:")
show_room(room)
print()
for i in range(len(room)):
    room[i]=clean_spot(room[i])
    print("after cleaning spot number")
    show_room(room)
print()
```
