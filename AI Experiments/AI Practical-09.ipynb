```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
df=pd.read_csv("ecommerce.csv")
df.head(10)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df_clean=df.dropna()
df_clean.shape
```

```python
df_clean['category'].unique()
```

```python
df_clean['category'].value_counts()
```

```python
X=df_clean['product_description']
y=df_clean['category']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state= 42,stratify = y)
```

```python
vectorizer  = CountVectorizer()
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
model = MultinomialNB()
model.fit(X_train_vec, y_train)
```

```python
new_reviews = [
    "The professor explained every single concept very clearly.",
    "Exam were unreasonable long and grading was completly un fair.",
    "Decent lectures, matched standard course content."
]
new_reviews_vec = vectorizer.transform(new_reviews)
predictions = model.predict(new_reviews_vec)
```

```python
for review, pred in zip(new_reviews, predictions):
    print(f"Review: '{review}' --> Predicted Sentiment: {pred}" )
```

```python
from sklearn.metrics import classification_report
y_pred = model.predict(X_test_vec)
# Classification Report
print(classification_report(y_test, y_pred))
```
